```python
import pandas  as pd
import numpy as np     
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_job_market_data.csv")
df.head()
```

```python
top_roles = df['title'].value_counts().head(10)

plt.figure(figsize=(10,5))
top_roles.plot(kind='bar')
plt.title("Top 10 Most Demanded Job Roles")
plt.xlabel("Job Role")
plt.ylabel("Count")
plt.show()
```

```python
#Experience Level Analysis

exp = df['formatted_experience_level'].value_counts()

plt.figure(figsize=(7,7``))
exp.plot(kind='pie', autopct='%1.1f%%')
plt.title("Experience Level Distribution")
plt.show()
```

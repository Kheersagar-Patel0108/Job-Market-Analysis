```python
import pandas as pd
df = pd.read_csv("job_market_data.csv", encoding='latin1')
df.head()

# remove duplicates rows and columns that are not useful for analysis

df.drop_duplicates(inplace=True)

columns_to_drop = ["company_url", "skills_desc","med_salary ","closed_time", 
    "id", "posting_domain_filter", "domain_tags"]

df.drop(columns=columns_to_drop, inplace=True, errors='ignore')
print(df.columns)

df.drop(columns=[
'application_url',
'posting_domain',
], inplace=True)
df = df.dropna(subset=['company_id'])

print(df.shape)
```

```python
# Handle Missing Values
#salary columns fill missing values with median, 

df['min_salary'] = df['min_salary'].fillna(df['min_salary'].median())
df['max_salary'] = df['max_salary'].fillna(df['max_salary'].median())


#Currency missing fill 

df['currency'] = df['currency'].fillna("Unknown")

#Experience missing fill

df['formatted_experience_level'] = df['formatted_experience_level'].fillna("Not Specified")

#Work type fill

df['work_type'] = df['work_type'].fillna("Not Specified")
df['remote_allowed'] = df['remote_allowed'].fillna("Unknown")
```

```python
#Create a Clean Salary Column
  #Average salary
df['average_salary'] = (df['min_salary'] + df['max_salary']) / 2
```

```python
#Convert Date Column into Proper Format

df['listed_time'] = pd.to_datetime(df['listed_time'], errors='coerce')
df['year']= df['listed_time'].dt.year
```

```python
df.info()
df.isnull().sum()

df.to_csv("cleaned_job_market_data.csv", index=False)
```
